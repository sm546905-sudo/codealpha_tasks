The Diwali Sales dataset was analyzed using Python, Pandas, Matplotlib and Seaborn. The data was cleaned and analyzed to understand customer characteristics and sales patterns.

Different factors such as gender, age group, state, occupation, marital status, zone and product category were analyzed. Charts were used to make the results easier to understand.

This project demonstrates the basic process of Exploratory Data Analysis (EDA), including data cleaning, analysis and visualization. The analysis helps in understanding customer purchasing patterns and sales distribution.
